```python
import json
import matplotlib.pyplot as plt

# Load histories
with open("history_train.json", "r") as f:
    history_train = json.load(f)

with open("history_val.json", "r") as f:
    history_val = json.load(f)

# Extract total loss
train_loss = [epoch[0] for epoch in history_train]
val_loss = [epoch[0] for epoch in history_val]
#
n_train = 1376
n_val = 688

train_mae = [x / n_train for x in train_loss]
val_mae = [x / n_val for x in val_loss]

epochs = range(len(train_mae))

# Plot
plt.figure()
plt.plot(epochs, train_mae, label="Train MAE")
plt.plot(epochs, val_mae, label="Validation MAE")
plt.xlabel("Epoch")
plt.ylabel("Mean Absolute Error (ASA)")
plt.title("Training vs Validation MAE")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Hardcoded from the training log
n_train    = 1376
n_val      = 688
batch_size = 4

# (total_loss = graph_loss, since all other losses are 0)
train_loss_per_epoch = [837610.5088, 593711.8599, 376899.8617, 327380.6435, 308490.7686]
val_loss_per_epoch   = [347738.6545, 221209.6967, 167225.1715, 157491.5459, 151709.6481]

train_mae = [(l * batch_size) / n_train for l in train_loss_per_epoch]
val_mae   = [(l * batch_size) / n_val   for l in val_loss_per_epoch]
epochs    = list(range(len(train_mae)))

best_epoch = val_mae.index(min(val_mae))

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(epochs, train_mae, label='Train MAE', marker='o', markersize=3, linewidth=2)
ax.plot(epochs, val_mae,   label='Val MAE',   marker='s', markersize=3, linewidth=2)
ax.axvline(best_epoch, color='gray', linestyle='--', linewidth=1,
           label=f'Best epoch {best_epoch} ({min(val_mae):.1f} m²/g)')
ax.set_xlabel('Epoch', fontsize=12)
ax.set_ylabel('MAE (m²/g)', fontsize=12)
ax.set_title('ALIGNN Baseline (no pore features)', fontsize=12)
ax.legend(fontsize=10, loc='upper right')
ax.grid(True, alpha=0.3)
plt.tight_layout()

save_path = '/home/milan/Downloads/alignn/Thesis/baseline_mae_curve.png'
plt.savefig(save_path, dpi=150)
plt.show()

print(f'Saved: {save_path}')
print(f'Best val MAE: {min(val_mae):.2f} m²/g at epoch {best_epoch}')
```

```python
import torch
from alignn.data import get_train_val_loaders
from alignn.config import TrainingConfig
from alignn.models.alignn import ALIGNN


# STORAGE FOR EMBEDDINGS
embeddings = {}

def save_hook(name):
    def hook(module, inp, out):
        # case 1: single tensor
        if torch.is_tensor(out):
            embeddings[name] = out.detach().cpu()

        # case 2: tuple/list output (ALIGNN does this often)
        elif isinstance(out, (tuple, list)):
            embeddings[name] = [
                o.detach().cpu() if torch.is_tensor(o) else o
                for o in out
            ]

        else:
            embeddings[name] = out
    return hook

# 4. ATTACH HOOKS (IMPORTANT)

for name, module in net.named_modules():
    n = name.lower()

    # atom-level embeddings
    if "atom" in n:
        module.register_forward_hook(save_hook(name))

    # bond / edge / line graph embeddings
    if "edge" in n or "line" in n:
        module.register_forward_hook(save_hook(name))

    # final representation before output
    if "dense" in n or "mlp" in n:
        module.register_forward_hook(save_hook(name))

print("Hooks registered")

# =========================
# 5. RUN ONE BATCH
# =========================
batch = next(iter(train_loader))
(graph, target) = prepare_batch(batch)

g, lg = graph

# ALIGNN still requires lattice tensor (even if unused)
lat = torch.zeros((1, 3, 3))

with torch.no_grad():
    pred = net((g, lg, lat))

print("Prediction:", pred)
print("Target:", target)

# =========================
# 6. CHECK EMBEDDINGS
# =========================
print("\nCaptured embeddings keys:")
for k in embeddings.keys():
    print("-", k)

# Example access:
# embeddings["<layer_name>"].shape
```
